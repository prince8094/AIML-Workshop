```python
df['Age'] = 2014 - df['Year_Birth']
```

```python
df['Total_Spending'] = (
    df['MntWines'] +
    df['MntFruits'] +
    df['MntMeatProducts'] +
    df['MntFishProducts'] +
    df['MntSweetProducts'] +
    df['MntGoldProds']
)
```

```python
df['Total_Orders'] = (
    df['NumWebPurchases'] +
    df['NumCatalogPurchases'] +
    df['NumStorePurchases']
)
```

```python
df['Average_Order_Value'] = (
    df['Total_Spending'] / df['Total_Orders'].replace(0, np.nan)
)
```

```python
df['Average_Order_Value'] = df['Average_Order_Value'].fillna(0)
```

```python
df['Purchase_Frequency'] = df['Total_Orders']
```

```python
df['Discount_Usage'] = df['NumDealsPurchases']
```

```python
df['Days_Since_Last_Purchase'] = df['Recency']
```

```python
df = df.drop([
    'ID',
    'Year_Birth',
    'Dt_Customer'
], axis=1)
```

```python
features = [
    'Age',
    'Income',
    'Total_Spending',
    'Total_Orders',
    'Average_Order_Value',
    'Purchase_Frequency',
    'Discount_Usage',
    'Days_Since_Last_Purchase'
]
```

```python
X = df[features]
```

```python
X.head()
```

```python
df = pd.get_dummies(
    df,
    columns=['Education', 'Marital_Status'],
    drop_first=True
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
```

```python
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(2, 11), wcss, marker='o')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method')

plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    print("K =", k, "Silhouette Score =", score)
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)
```

```python
df['Cluster'] = cluster_labels
```

```python
df.head()
```

```python
df['Cluster'].value_counts().sort_index()
```

```python
sns.scatterplot(
    data=df,
    x='Income',
    y='Total_Spending',
    hue='Cluster'
)

plt.title('Income vs Total Spending')
plt.xlabel('Annual Income')
plt.ylabel('Total Spending')

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Purchase_Frequency',
    y='Average_Order_Value',
    hue='Cluster'
)

plt.title('Purchase Frequency vs Average Order Value')
plt.xlabel('Purchase Frequency')
plt.ylabel('Average Order Value')

plt.show()
```

```python
kmeans.cluster_centers_
```

```python
centroids = scaler.inverse_transform(
    kmeans.cluster_centers_
)
```

```python
centroids_df = pd.DataFrame(
    centroids,
    columns=features
)
```

```python
centroids_df
```

```python
plt.scatter(
    df['Income'],
    df['Total_Spending'],
    c=df['Cluster']
)

plt.scatter(
    centroids_df['Income'],
    centroids_df['Total_Spending'],
    marker='X',
    s=200
)

plt.xlabel('Annual Income')
plt.ylabel('Total Spending')
plt.title('Clusters with Centroids')

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Cluster'
)

plt.title('Customer Distribution by Cluster')
plt.xlabel('Cluster')
plt.ylabel('Number of Customers')

plt.show()
```
