```python
df['Product_Category'] = df['Product_Category'].replace({
    'home & kitchen': 'Home & Kitchen',
    'Home and Kitchen': 'Home & Kitchen',
    'electronics': 'Electronics',
    'Electronic': 'Electronics',
    'grocery': 'Beauty',
    'beauty': 'Patna',
    'Clothing': 'Clothes',
    'books' : 'Books',
    'clothing': 'Clothes'
})
```

```python
#remove invalid age using reasonabe bussinus rule
df = df[(df['Age'].between(18, 65)) | (df['Age'].isna())]
```

```python
#Handle the missing ages separately
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python


print("Unique Age values after cleaning:")
print(df['Age'].unique())

print("\nMissing values after cleaning:")
print(df[['Age']].isnull().sum())
```
